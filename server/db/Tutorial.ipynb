```python
from manager import MongoDBAtlasManager
from datetime import datetime
```

```python
# manager 생성
db = MongoDBAtlasManager()
```

```python
# CREATE
print("[CREATE]")
id1 = db.create(
    created_at=datetime(2026, 1, 22, 10, 30),
    remote_file_path="s3://bucket/image_001.dat",
    location_name="백령도",
    metadata={"camera": "CCTV"}
)
id2 = db.create(
    created_at=datetime(2026, 1, 22, 14, 15),
    remote_file_path="s3://bucket/image_002.dat",
    location_name="연평도",
    metadata={"camera": "아이폰"}
)
```

```python
# READ
print("[READ - 전체]")
all_data = db.read_all()
for doc in all_data:
    print(f"  {doc['remote_file_path']} \ {doc['location_name']} ")

print("\n[READ - 시간 범위]")
time_data = db.read_by_time(
    datetime(2026, 1, 22, 0, 0),
    datetime(2026, 1, 22, 23, 59)
)
print(f"  {len(time_data)}개 파일")

print("\n[READ - 위치]")
location_data = db.read_by_location("Seoul")
print(f"  Seoul: {len(location_data)}개 파일")
```

```python

# DELETE
print("\n[DELETE]")
deleted = db.delete_by_path("s3://bucket/image_0001.dat")
print(f"  삭제된 문서: {deleted}개")

```
